## Exercise 1: Numeric analysis

- Create one `int`, one `float` (scientific notation), and one `complex` number.
- Print each value, type, and (for complex) real/imag parts.
- Convert int->float and float->int, then print results.

```python
# Write your solution here
# 1. Create one int, one float (scientific notation), and one complex number
my_int = 7
my_float = 4.5e3  # Equivalent to 4500.0
my_complex = 3 + 4j

# 2. Print each value, type, and (for complex) real/imag parts
print("--- Value and Type Analysis ---")
print(f"Int Value: {my_int}, Type: {type(my_int)}")
print(f"Float Value: {my_float}, Type: {type(my_float)}")
print(f"Complex Value: {my_complex}, Type: {type(my_complex)}")
print(f"Complex Real Part: {my_complex.real}, Imaginary Part: {my_complex.imag}")

# 3. Convert int->float and float->int, then print results
int_to_float = float(my_int)
float_to_int = int(my_float)

print("\n--- Type Casting Results ---")
print(f"Converted Int to Float: {int_to_float}, Type: {type(int_to_float)}")
print(f"Converted Float to Int: {float_to_int}, Type: {type(float_to_int)}")
```

## Exercise 2: Safe casting challenge

- Given `values = ["25", "3.14", "-8", "hello"]`, decide how each text should be converted.
- Use string methods (such as `isdigit()`, `replace()`, `lstrip()`) and `if/elif/else` to check each value.
- Convert to `int` if it is a whole number, to `float` if it is a decimal number, otherwise keep it as a string.
- Print each converted value together with its type.

```python
# Write your solution here
values = ["25", "3.14", "-8", "hello"]

print("--- Safe Casting Results ---")

for text in values:
    # 1. Clean the text for negative sign checks
    clean_text = text.lstrip('-')
    
    # 2. Check for whole numbers (integers)
    if clean_text.isdigit():
        converted = int(text)
        
    # 3. Check for decimal numbers (floats)
    # We replace the first occurrence of a decimal point to see if the rest is digits
    elif '.' in text and text.replace('.', '', 1).lstrip('-').isdigit():
        converted = float(text)
        
    # 4. Fallback to original string if it doesn't match numeric criteria
    else:
        converted = text
        
    print(f"Original: '{text}' -> Converted: {repr(converted)}, Type: {type(converted)}")
```

## Exercise 3: String slicing matrix

- Use `text = "DataSciencePython"`.
- Print slices: first 4 chars, last 6 chars, every 2nd char, reversed text.
- Extract and print `Science` from the string using slicing only.

```python
# Write your solution here
text = "DataSciencePython"

# 1. Print standard slices
first_four = text[:4]
last_six = text[-6:]
every_second = text[::2]
reversed_text = text[::-1]

print("--- Matrix Slicing Results ---")
print(f"First 4 characters:  '{first_four}'")
print(f"Last 6 characters:   '{last_six}'")
print(f"Every 2nd character: '{every_second}'")
print(f"Reversed text:       '{reversed_text}'")

# 2. Extract and print "Science" using slicing only
# "Science" starts at index 4 and ends before index 11
science_extract = text[4:11]
print(f"\nExtracted Substring: '{science_extract}'")
```

## Exercise 4: String transformation pipeline

- Start with `raw = "   python,java,c++ , go  "`.
- Strip the outer spaces and split the text by comma.
- Print the split list.
- Loop through the list, strip each language, uppercase it, and print it.

```python
# Write your solution here
# 1. Create raw string
raw = "   python,java,c++ , go  "

# 2. Strip outer spaces and split by comma
languages = raw.strip().split(",")

# 3. Print the split list
print(languages)

# 4. Loop through the list
for language in languages:
    language = language.strip()
    language = language.upper()
    print(language)
```

## Exercise 5: Formatting and escaping

- Create variables `product`, `price`, and `discount`.
- Print original and discounted price using f-string with 2 decimals.
- Print one line containing escaped quotes and another with newline+tab formatting.

```python
# Write your solution here
# 1. Create variables
product = "Laptop"
price = 850.50
discount = 10

# 2. Calculate discounted price
discounted_price = price - (price * discount / 100)

# 3. Print original and discounted price using f-string
print(f"Product: {product}")
print(f"Original Price: ${price:.2f}")
print(f"Discount: {discount}%")
print(f"Discounted Price: ${discounted_price:.2f}")

# 4. Escaped quotes
print("The product name is \"Laptop\".")

# 5. Newline and tab formatting
print("Product Information:\n\tName: Laptop\n\tPrice: $850.50")
```
