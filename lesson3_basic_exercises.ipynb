## Exercise 3: Logical operators

1. Create `age = 20` and `has_id = True`.
2. Use `and` to check if someone can enter: age >= 18 and has_id is True.
3. Use `or` to check if `age < 18` or `has_id` is False.
4. Use `not` with `has_id` and print the result.

```python
# Write your solution here
# 1. Create variables
age = 20
has_id = True

# 2. Use AND operator
can_enter = age >= 18 and has_id is True
print("Can enter:", can_enter)

# 3. Use OR operator
check = age < 18 or has_id is False
print("Underage or no ID:", check)

# 4. Use NOT operator
print("Does not have ID:", not has_id)
```

## Exercise 4: Identity and membership

1. Create two lists:
   - `list1 = [1, 2, 3]`
   - `list2 = list1`
2. Print `list1 is list2` and `list1 is not list2`.
3. Check membership in `list1`:
   - `2 in list1`
   - `5 not in list1`

```python
# Write your solution here
# 1. Create two lists
list1 = [1, 2, 3]
list2 = list1

# 2. Identity operators
print(list1 is list2)
print(list1 is not list2)

# 3. Membership operators
print(2 in list1)
print(5 not in list1)
```

## Exercise 5: Arithmetic and assignment

1. Create `n = 10`.
2. Update `n` using these operators in order and print after each step:
   - `n += 5`
   - `n -= 3`
   - `n *= 2`
   - `n /= 4`
3. Print `n % 3` and `n ** 2`.

```python
# Write your solution here
# 1. Create variable
n = 10

# 2. Assignment operators
n += 5
print("After += 5:", n)

n -= 3
print("After -= 3:", n)

n *= 2
print("After *= 2:", n)

n /= 4
print("After /= 4:", n)

# 3. Arithmetic operators
print("n % 3:", n % 3)
print("n ** 2:", n ** 2)
```
