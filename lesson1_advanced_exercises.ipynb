## Exercise 1: Variable swap and trace

- Create `a = 15` and `b = 40`.
- Swap them **without** using a third variable.
- Print before/after values in a clear format.
- Print the type of each variable after swapping.

```python
# 1. Initialize the variables
a = 15
b = 40

# 2. Print before values
print("--- Before Swap ---")
print(f"a = {a}")
print(f"b = {b}")

# 3. Swap variables without a third variable (Tuple Unpacking)
a, b = b, a

# 4. Print after values
print("\n--- After Swap ---")
print(f"a = {a}")
print(f"b = {b}")

# 5. Print the type of each variable after swapping
print("\n--- Variable Types ---")
print(f"Type of a: {type(a)}")
print(f"Type of b: {type(b)}")
```

## Exercise 3: Parallel assignment and unpacking

- Assign `x, y, z` in one line with three different data types.
- Assign the same value to `a`, `b`, and `c` in one line.
- Create a list of 3 values and unpack it into three variables.
- Print all variables and the type of `x`, `y`, and `z`.

```python
# 1. Parallel Assignment
x, y, z = "Python", 100, 10.5

# 2. Multiple Assignment
a = b = c = 50

# 3. List Unpacking
numbers = [10, 20, 30]
num1, num2, num3 = numbers

# 4. Print all variables
print("x =", x)
print("y =", y)
print("z =", z)

print("a =", a)
print("b =", b)
print("c =", c)

print("num1 =", num1)
print("num2 =", num2)
print("num3 =", num3)

# 5. Print data types of x, y, z
print(type(x))
print(type(y))
print(type(z))
```

## Exercise 5: Type-driven expression building

- Create variables for name (str), age (int), height (float), active (bool).
- Print all four variables in one line using commas inside `print()`.
- Create `first_name` and `last_name`, then print the full name using `+` and a space.
- Add `age` and `height`, print the result and its type.
- Print the type of every variable at the end.

```python
# 1. Create variables
name = "John"
age = 23
height = 1.75
active = True

# 2. Print all variables in one line
print(name, age, height, active)

# 3. Combine first name and last name
first_name = "John"
last_name = "Smith"

full_name = first_name + " " + last_name
print(full_name)

# 4. Add age and height
result = age + height

print(result)
print(type(result))

# 5. Print the type of every variable
print(type(name))
print(type(age))
print(type(height))
print(type(active))
print(type(first_name))
print(type(last_name))
print(type(full_name))
print(type(result))
```
