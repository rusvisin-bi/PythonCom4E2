## Exercise 1: Truth table mini-lab

- Create booleans `p = True`, `q = False`.
- Print results for: `p and q`, `p or q`, `not p`, `not q`.
- Add one compound expression with parentheses and print its result.

```python
# Write your solution here
# 1. Create boolean variables
p = True
q = False

# 2. Logical operators
print("p and q:", p and q)
print("p or q:", p or q)
print("not p:", not p)
print("not q:", not q)

# 3. Compound expression
result = (p and q) or (not q)
print("Compound expression:", result)
```

## Exercise 2: Comparison chain

- Given `a = 12`, `b = 12`, `c = 7`, print at least 6 comparison results.
- Include equality, inequality, greater/less, and a chained comparison (e.g., `c < a <= b`).

```python
# Write your solution here
# 1. Create variables
a = 12
b = 12
c = 7

# 2. Comparison operators
print("a == b:", a == b)
print("a != c:", a != c)
print("a > c:", a > c)
print("c < a:", c < a)
print("a >= b:", a >= b)
print("c <= b:", c <= b)

# 3. Chained comparison
print("c < a <= b:", c < a <= b)
```

## Exercise 3: Operator precedence check

- Create a complex boolean expression using `and`, `or`, and `not`.
- Evaluate it once without extra parentheses and once with explicit parentheses.
- Print both and comment whether they match.

```python
# Write your solution here
# 1. Create boolean variables
p = True
q = False
r = True

# 2. Expression without extra parentheses
result1 = p or q and not r

# 3. Expression with explicit parentheses
result2 = (p or q) and (not r)

# 4. Print results
print("Without parentheses:", result1)
print("With parentheses:", result2)

# 5. Compare both results
print("Do they match?", result1 == result2)

# They do not match because operator precedence is different.
```

## Exercise 4: Identity vs equality

- Create two equal lists and compare with `==` and `is`.
- Assign one list to another variable and compare again.
- Write a short comment explaining differences in output.

```python
# Write your solution here
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print("list1 == list2:", list1 == list2)
print("list1 is list2:", list1 is list2)

list3 = list1

print("list1 == list3:", list1 == list3)
print("list1 is list3:", list1 is list3)
```

## Exercise 5: Membership filter

- Given `items = ["python", "java", "go", "rust"]`, test membership for 4 queries.
- Use both `in` and `not in`.
- Print human-readable results like `python: found`.

```python
# Write your solution here
# 1. Create a list
items = ["python", "java", "go", "rust"]

# 2. Test membership using in
print("python:", "found" if "python" in items else "not found")
print("java:", "found" if "java" in items else "not found")

# 3. Test membership using not in
print("javascript:", "not found" if "javascript" not in items else "found")
print("php:", "not found" if "php" not in items else "found")
```
